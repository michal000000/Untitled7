```python
import PyMPDATA as mp

from PyMPDATA import ScalarField as sf, VectorField as vf, Options as op, boundary_conditions as bc, Stepper as stepa, Solver as solva

import numpy as np
import numba as nb

import matplotlib.pyplot as plt
```

# 1

```python

print(mp.__file__)

op = op(n_iters=1)

common = {'halo': op.n_halo, 'boundary_conditions': (bc.Periodic(), )}

psi = sf( **common, data=np.exp( -(x := np.linspace(-2, 8, 40))*x ) )

C = vf( **common, data= (np.full(x.size+1, 0.25), ) )

plt.step(x, psi.get() )
```

```python

solver = solva(advector=C, advectee=psi, stepper=stepa(n_dims=1, options=op) )

solver.advance(500)

plt.step(x, psi.get() )
```
